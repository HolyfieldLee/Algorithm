# 📝 문제 풀이 핵심

큰 동전부터 최대한 사용한다.

나머지를 갱신하며 반복하면 최소 개수 보장.

# 📝English, Core Idea

Use the largest coin first as much as possible.

Update the remaining amount until it becomes zero, ensuring the minimum number of coins.

```python
N,K = map(int, input().split())
A = [0]*N

for i in range(N):
    A[i] = int(input())

count = 0

for i in range(N - 1, -1, -1):
    if A[i] <= K:                #현재 동전의 가치가 K보다 작거나 같으면 구성에 추가!
        count += int(K / A[i])
        K = K % A[i]

print(count)
```
