# 📌 문제 요약

* 여러 장의 카드 묶음이 있음.

* 두 묶음을 합칠 때는 두 묶음 크기의 합만큼 비용이 듦.

* 최종적으로 모든 묶음을 합칠 때, 비용 합의 최솟값을 구하는 문제.

* 👉 핵심 아이디어: 항상 제일 작은 두 묶음을 먼저 합쳐야 전체 비용이 최소가 됨 (그리디 전략).\

* ### Tip: 일반 큐는 먼저 들어온 게 먼저 나오지만, 우선순위 큐는 가장 작은 값이 먼저 나옴.

# 📌 Problem Summary

* You have several piles of cards.

* When combining two piles, the cost is equal to the sum of their sizes.

* The goal is to find the minimum total cost required to combine all piles into one.

* 👉 Key idea: Always combine the two smallest piles first to minimize the overall cost (Greedy strategy).

* ### Tip: Unlike a regular queue (FIFO: first in, first out), a priority queue always pops the smallest element first.

```python
from queue import PriorityQueue

N = int(input())
pq = PriorityQueue()

for _ in range(N):
    date = int(input())
    pq.put(date)

data1 = 0
data2 = 0
sum = 0

while pq.qsize() > 1:
    data1 = pq.get()
    data2 = pq.get()
    temp = data1 + data2
    sum += temp
    pq.put(temp)

print(sum)
```
