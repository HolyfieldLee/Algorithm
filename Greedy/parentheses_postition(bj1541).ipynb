# 📌 전체 흐름 요약

* '-' 기준으로 잘라서 리스트에 저장.

* 각 조각 안에서 +를 먼저 계산해서 합계 구함.

* 첫 번째 조각은 더하고, 나머지는 전부 빼줌.

* 결과 출력 → 최소값 완성 ✅

* 그리디 문제
### 👉 항상 -가 등장한 이후는 전부 괄호로 묶는다.

# 📌 Overall Flow Summary

* Split the expression by '-' and store it in a list.

* For each part, calculate the sum of numbers separated by '+'.

* Add the first part, and subtract all the others.

* Print the result → this gives the minimum value ✅

* Greedy Problem

### 👉 Once a '-' appears, treat everything after it as if it’s inside parentheses.

```python
answer = 0
A = list(map(str, input().split("-")))

def mySum(i):
    sum = 0
    temp = str(i).split("+")
    for i in temp:
        sum += int(i)
    return sum

for i in range(len(A)):
    temp = mySum(A[i])
    if i == 0:
        answer += temp # 가장 앞에 있는 값만 더하기
    else:
        answer -= temp # 뒷부분의 값은 합쳐서 빼기

print(answer)
```
