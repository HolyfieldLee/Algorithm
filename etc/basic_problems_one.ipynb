# ✅ 문제 1. 체육복 문제

학생들이 번호순(1번 ~ n번)으로 줄을 서 있어요.

일부 학생은 체육복을 잃어버림. (lost 배열)

일부 학생은 체육복을 여분으로 가지고 있음. (reserve 배열)

체육복은 자기 자신 또는 바로 앞/뒤 번호 학생에게만 빌려줄 수 있어요.

최대로 수업에 참여할 수 있는 학생 수를 구하라.

🔹 문제 상황 예시

학생 수 = 5명
잃어버린 학생 = [2, 4]
여분 있는 학생 = [3]

👉 이 경우,

2번은 체육복 없음

3번은 여분이 있으니 2번에게 빌려줌

4번은 빌릴 수 있는 3번이 이미 체육복 다 줬으니 빌릴 수 없음

➡️ 총 4명만 수업 참여 가능

📌 문제 접근 방법

겹치는 경우 제거

어떤 학생이 lost에도 있고 reserve에도 있을 수 있어요.

이 경우, 그 학생은 그냥 자기 옷만 입는다고 생각하고 제외해야 합니다.

앞뒤 학생에게 빌려주기

작은 번호부터 차례대로 확인하면서,

lost 학생이 있으면

앞번호(i-1)에 여분이 있으면 빌리고,

아니면 뒷번호(i+1)에 여분이 있으면 빌립니다.

최종 학생 수 계산

전체 학생 수 n에서 체육복 못 받은 학생 수를 빼주면 됩니다.

```python
def solution(n, lost, reserve):
    # 1. 겹치는 학생 정리
    lost_set = set(lost) - set(reserve)
    reserve_set = set(reserve) - set(lost)

    # 2. 체육복 빌려주기
    for r in sorted(reserve_set):  # 작은 번호부터 탐색
        if r - 1 in lost_set:      # 앞번호 학생이 체육복 없음
            lost_set.remove(r - 1)
        elif r + 1 in lost_set:    # 뒷번호 학생이 체육복 없음
            lost_set.remove(r + 1)

    # 3. 최종 참여 인원 = 전체 - 체육복 없는 학생
    return n - len(lost_set)
```

# ✅ 문제 2. 문자열 압축

문제 설명
연속된 문자의 개수를 세서 압축된 문자열로 반환합니다.
예) "aaabbccccdaa" → "a3b2c4d1a2"

풀이 아이디어

1. 문자열을 왼쪽부터 탐색하며 같은 문자가 연속되는 구간을 셉니다.

2. 문자가 바뀌면 문자+개수를 결과에 추가.

3. 마지막 문자도 추가.

```python
def compress_string(s):
    result = []
    count = 1
    
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result.append(s[i-1] + str(count))
            count = 1
    result.append(s[-1] + str(count))  # 마지막 문자 처리
    
    return ''.join(result)
```

# ✅ 문제 3. 최빈값 구하기

문제 설명
리스트 안에서 가장 많이 나온 숫자를 찾아야 합니다. 만약 빈도가 같은 숫자가 여러 개라면 가장 작은 수를 반환합니다.

풀이 아이디어

1. 각 숫자가 몇 번 나왔는지 세어야 합니다. → collections.Counter 사용 가능.

2. 가장 많이 나온 횟수를 찾습니다.

3. 그 횟수에 해당하는 숫자 중 가장 작은 수를 고릅니다.

```python
from collections import Counter

def mode_value(nums):
    counter = Counter(nums)                  # 등장 횟수 세기
    max_freq = max(counter.values())         # 최빈 빈도
    
    candidates = []                          # 후보 리스트 준비
    for num, cnt in counter.items():
        if cnt == max_freq:                  # 최빈 빈도와 같으면
            candidates.append(num)           # 후보에 추가
    
    return min(candidates)                   # 가장 작은 수 반환
```
