# 최단거리? => BFS!

그리고 동시에 이 문제에서 BFS를 떠올린 이유는 BFS는 가까운 노드부터 차례로 방문한다.
즉, 한단계씩 확장하기에 출발점에서 최단거리를 자연스럽게 구할 수 있다.
그리고 BFS를 떠올린다면, 가중치가 있냐 없냐도 자동으로 생각하자!

그리고 거리 저장용 리스트 visited도 만들어 두었다.

# Shortest Path? => BFS!
The reason I thought of BFS for this problem is that BFS visits the nearest nodes first in sequential order.
In other words, because it expands one level at a time, it can naturally find the shortest distance from the starting point.

Also, when you think of BFS, you should automatically consider whether the graph is weighted or unweighted.

I also created a visited list to store the distances.

```python
from collections import deque
N,M,K,X = map(int, input().split())
A = [[] for _ in range(N+1)]
answer = []
visited = [-1] * (N + 1)

def BFS(v):
    queue = deque()
    queue.append(v)
    visited[v] += 1
    # append 한 다음 1을 더한다. 공간에 넣고 visted의 v번째 인덱스에도 1값을 넣어 주는 것!
    while queue:
        now_Node = queue.popleft()
        for i in A[now_Node]:
            if visited[i] == -1:
                visited[i] = visited[now_Node] + 1
                queue.append(i)

for _ in range(M):
    S, E = map(int, input().split())
    A[S].append(E)

BFS(X)

for i in range(N + 1):
    if visited[i] == K:
        answer.append(i)

if not answer:
    print(-1)
else:
    answer.sort()
    for i in answer:
        print(i)
```
