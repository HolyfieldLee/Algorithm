## 1. 문제 핵심

N×N 배열: A[i][j] = i × j

배열을 정렬했을 때 K번째 수 찾기.

배열 크기가 너무 크므로 직접 생성 불가 → 이진 탐색 활용.

## 2. 이진 탐색 아이디어

값의 범위: 1 ~ N×N (혹은 1 ~ K).

mid = (start + end)//2로 중앙값 설정.

count(mid) = 배열 안에서 mid 이하 원소 개수.

## 3. count(mid) 계산법

i행 값: i, 2i, 3i, …, Ni

i×j ≤ mid ⇔ j ≤ mid//i

따라서 i행의 개수: min(N, mid//i)

전체: count(mid) = Σ min(N, mid//i)

## 4. 탐색 로직

count(mid) < K → 더 큰 값 탐색 → start = mid+1

count(mid) ≥ K → 후보 저장, 왼쪽 탐색 → ans = mid, end = mid-1

# 📘 🇺🇸 English Summary


## 1.  Problem Core

N×N table: A[i][j] = i × j

Find the K-th smallest number after sorting.

The table is too large to build → use binary search.

## 2. Binary Search Idea

Value range: 1 ~ N×N (or 1 ~ K).

Choose mid = (start + end)//2.

count(mid) = number of elements ≤ mid.

## 3. How to compute count(mid)

Row i values: i, 2i, 3i, …, Ni.

Condition: i×j ≤ mid ⇔ j ≤ mid//i.

Row i count: min(N, mid//i).

Total: count(mid) = Σ min(N, mid//i).

## 4. Search Logic

If count(mid) < K → answer is larger → start = mid+1.

If count(mid) ≥ K → store candidate, search smaller side → ans = mid, end = mid-1.

```python
N = int(input())
K = int(input())
start = 1
end = K
ans = 0 # 조건을 만족할 후보 정답의 임시 변수

while start <= end: # 탐색 구간이 남아 있는 동안 반복
    middle = int((start + end) / 2)
    cnt = 0
    for i in range(1, N+1):
        cnt += min(int(middle / i), N) # 작은 수 카운트 핵심 로직
    if cnt < K:
        start = middle + 1


# middle 이하가 K개 이상이면, K번째 수는 middle보다 크지 않을 수도 있습니다.
# → 현재 middle을 후보 정답으로 저장해 두고,
# → 더 작은 값 쪽에서도 답이 있는지 왼쪽 구간으로 좁혀봅니다.
# (이렇게 해야 “가장 작은 만족값”(lower bound)을 찾습니다.)
# end = middle - 1 추가설명 → 그런데 혹시 더 작은 값이 정답일 수도 있잖아요?
# → 그래서 탐색 범위를 왼쪽으로 좁히려고 end = middle - 1을 하는 겁니다.
    else:                              
        ans = middle
        end = middle - 1
print(ans)
```
