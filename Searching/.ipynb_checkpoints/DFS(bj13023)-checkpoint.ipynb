이 문제는 “연속된 5명이 친구 관계로 이어지는지”를 판별하는 문제인데, 여기서 핵심 단서는 **“연속성”과 “깊이”**에 있다. BFS처럼 최단 거리를 구하는 게 아니라, 특정 깊이까지 이어지는 단순 경로가 존재하는지를 확인해야 하므로 DFS 발상이 자연스럽게 이어진다. 문제 크기(N, M ≤ 2000)도 DFS 백트래킹으로 충분히 가능하며, 실제로는 “정점 5개(사람 5명)”가 이어지는지를 확인해야 한다. 따라서 DFS에서 depth=5가 되는 순간 정점 5개가 연속으로 연결된 것이므로 답을 찾을 수 있다. 여기서 중요한 포인트는 정점 5개가 연결되면 간선은 4개라는 점이다. 즉, DFS 깊이는 방문한 사람 수(정점 수)를 의미하고, 간선은 그보다 하나 적은 값이 된다.



This problem (Baekjoon 13023, ABCDE) asks whether there exists a sequence of five people connected through friendship relations. The key hints in the problem statement are “continuity” and “depth.” Instead of finding the shortest distance like BFS, we need to check whether a simple path of a certain depth exists, which naturally leads to using DFS. The input size (N, M ≤ 2000) also makes DFS with backtracking feasible. In practice, what we are checking is whether five vertices (five people) are connected consecutively. Therefore, when the DFS reaches depth = 5, it means five vertices are connected, so the condition is satisfied. An important detail is that if five vertices are connected, there will be four edges. In other words, the DFS depth represents the number of visited vertices, while the number of edges will always be one less.

```python
N, M = map(int, input().split())
arrive = True
A = [[] for _ in range(N + 1)]
visited = [False] * (N + 1)

def DFS(now, depth):
    global arrive
    if depth == 5:
        arrive = True
        return
    visited[now] = True
    for i in A[now]:
        if not visited[i]:
            DFS(i, depth + 1)
    visited[now] = False

for _ in range(M):
    s, e = map(int, input().split())
    A[s].append(e)
    A[e].append(s)

for i in range(N):
    DFS(i, 1)
    if arrive:
        break
if arrive:
    print(1)
else:
    print(0)
```
