# 이진탐색


* 찾을 때까지 반복해야 하므로 전체 과정을 for문 안에 두자. for i in range(M):
* find = False로 기본값 설정해두는건 꼭 잊지 말자
* target값은 target_list[인덱스]로 해야한다.
* target 값이 중간값보다 작으면, 당연히 end값을 현재 중간값보다도 작게 만들어야 한다. 그래서 end = midi - 1
* target 값이 중간값보다 크면, 당연히 start값을 현재 중간값보다도 크게 만들어야 한다. 그래서 start = midi + 1
*그리고 출력결과는 if-else를 사용해서 각각 print(1) , print(0) 으로 가볍게 코딩해주면 된다.

# Binary Search

* Since you need to keep searching until you find the value, put the whole process inside a for loop: for i in range(M):

* Don’t forget to set find = False as the default.

* The target value should be target_list[index].

* If the target value is smaller than the middle value, naturally you must also make the end smaller than the current middle index: end = midi - 1.

* If the target value is larger than the middle value, naturally you must make the start larger than the current middle index: start = midi + 1.

* For the output, simply use if-else with print(1) or print(0) for light and clean coding.

```python
N = int(input())
A = list(map(int, input().split()))
A.sort()
M = int(input())
target_list = list(map(int, input().split()))

for i in range(M):
    find = False
    target = target_list[i]

    start = 0
    end = len(A) - 1
    while start <= end:
        midi = int((start + end) / 2)
        midv = A[midi]
        if midv > target:
            end = midi - 1
        elif midv < target:
            start = midi + 1
        else:
            find = True
            break

    if find:
        print(1)
    else:
        print(0)
         
```
