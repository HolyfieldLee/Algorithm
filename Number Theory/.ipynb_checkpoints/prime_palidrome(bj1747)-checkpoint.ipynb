# 1. 먼저 에라토스테네스로 소수들을 구한다
# 2. 그 중에서 펠린드롬인 것들을 찾으면 되는데 그 중에서 "최초로" 찾은 것을 선택하면 된다.
# 3. 펠린드롬은 start와 end 변수를 세팅하고 while문 안에서 s < e인 조건동안 모든 값이 같으면 펠린드롬으로 판별한다.

# 1. First, use the Sieve of Eratosthenes to find all the prime numbers within a given range. (In other words, eliminate composite numbers step by step, and the remaining numbers are primes.)

# 2. Among those primes, find the ones that are palindromes (numbers that read the same forward and backward). If multiple palindrome primes exist, choose the first one found — that is, the smallest palindrome prime.

# 3. To check whether a number is a palindrome:

    a. Set two pointers, start and end, at the beginning and end of the string (or number).

    b. While start < end, compare the characters at both positions.

    c. If all corresponding pairs are equal, it’s a palindrome; if any pair differs, it’s not.

So in short:

“Find all prime numbers, then pick the first one that’s also a palindrome.”

```python
import math
N = int(input())
A = [0]*(10000001)

for i in range(2, len(A)):
    A[i] = i                 # 인덱스를 자기 값으로 초기화

for i in range(2, int(math.sqrt(len(A)) + 1)): # 제곱근까지만 수행
    if A[i] == 0:
        continue
    for j in range(i + i, len(A), i): # 배수 지우기
        A[j] = 0

def isPalindrome(target): # 펠린드롬 수 판별 함수
    temp = list(str(target))
    s = 0
    e = len(temp) - 1
    while (s < e):
        if temp[s] != temp[e]:
            return False
        s += 1
        e -= 1

    return True

i = N
while True: # N부터 1씩 증가시키면서 소수와 펠린드롬 수가 맞는지 판별
    if A[i] != 0:
        result = A[i]
        if (isPalindrome(result)):
            print(result)
            break

    i += 1
```
