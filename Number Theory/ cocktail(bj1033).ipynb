## 🧩 이 문제를 보고 어떻게 최소공배, 최대공약, DFS를 떠올리는지 그 발상 메커니즘 알아내는 것이 관건!

### 1️⃣ 이 문제는 결국 무엇을 묻는 문제인가?

N개의 재료가 들어간 칵테일이 있다.
두 재료 사이의 비율(p:q)가 주어진다.
그런데 어떤 재료끼리는 직접적인 비율이 안 주어질 수도 있다.
=> 하지만 연결관계를 통해 모든 재료의 비율을 구해야 한다.

---

#### 즉, 모든 재료의 비율을 정수비로 표현하라.

재료 3개 (0, 1, 2)

0 : 1 = 2 : 
3
1 : 2 = 4 : 
0 / 1 = 2 / 3  



→ 0:1 비율은 2:3

1 / 2 = 4 / 5

---

이걸 하나의 전체 비율로 바꾸려면
각 재료가 같은 기준 단위(공통 배수) 에서 비교돼야 해요.

즉, 0과 1의 비율(2:3)을 기준으로 1과 2의 비율(4:5)를 “연결”해서
0:1:2 의 비율을 “연쇄적으로” 구해야 해요.

그래서 이 문제는 결국 "비율 그래프"를 순회하면서 전체 비율을 계산하는 문제입니다.
5


#### 🔹 (1) DFS를 떠올린 이유 – 관계가 연결 그래프 구조이기 때문

비율이 주어지는 형태는 “a와 b 사이의 관계”입니다.

즉,

a - b 사이에 (p:q)라는 간선이 존재

→ 이건 전형적인 그래프의 간선 정보입니다.

즉,
각 재료를 노드로,
비율을 간선으로 연결하면
비율 관계가 트리 형태로 만들어져요.
(입력 조건이 N-1개니까, 트리 구조임 ✅)

👉 그래서 이 비율 트리를 DFS로 순회하면서 각 노드의 실제 비율값을 누적해가는 거예요.

```python
N = int(input())
A = [[] for _ in range(N)]
visited = [False] * (N)
D = [0] * N
lcm = 1

def gcd(a,b):
    if b == 0:
        return a
    else:
        return gcd(b, a % b)

def DFS(v):                       #DFS 탐색 함수 구현
    visited[v] = True
    for i in A[v]:
        next = i[0]
        if not visited[next]:
            D[next] = D[v] * i[2] // i[1]
            DFS(next)

for i in range(N - 1):
    a,b,p,q = map(int, input().split())
    A[a].append((b,p,q))
    A[b].append((a,p,q))
    lcm *= (p * q // gcd(p,q)) # 최소 공배수는 두 수의 곱을 최대 공약수로 나눈 것

D[0] = lcm
DFS(0)
mgcd = D[0]

for i in range(1 ,N):
    mgcd = gcd(mgcd, D[i])

for i in range(N):
    print(int(D[i] // mgcd), end= ' ')
```
